```python
df=spark.read.option("multiline",True).json("/Volumes/workspace/default/json/orders_nested.json")
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
df_customer=df.select(col("customer.contact").alias("cc"),col("customer.customer_id").alias("customer_id"),col("customer.customer_name").alias("customer_name"),col("customer.loyalty").alias("customer_loyalty"))
```

```python
df_customer=df_customer.select("customer_id","customer_name","customer_loyalty",col("cc.email").alias("email"),col("cc.phone").alias("phone"))
```

```python
df_customer.show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col,explode
df_items= df.select(col("customer.customer_id").alias("customer_id"),explode(col("items")).alias("items"))
```

```python
df_items.printSchema()
```

```python
df_items=df_items.select(col("customer_id"),col("items.category").alias("category"),col("items.discount.amount_off").alias("amount"),col("items.discount.percent_off").alias("percent_off"),col("items.product_id").alias("product_id"),col("items.unit_price").alias("unit_price"),col("items.quantity").alias("quantity"),col("items.product_name").alias("product_name"))
```
